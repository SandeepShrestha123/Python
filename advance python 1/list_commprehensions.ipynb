```python
# list Comprehension is a tehcnique to create a new list with the help of existing list 
#using a single line of code. Instead of writing a for loop and using append(), you can write everything in one line.
```

```python
# creating a new list to store the square of the numbers from the existing list using list comprehensions

number = [2,3,4,5,6]
print(number)
square_number = [i ** 2 for i in range(1, 6)]
print(square_number)
```

```python
# converting a list name to upperecase using list comprehensions

names = ["sandeep", "srestaa", "anup", "ramesh"]
print(names)
cap_names = [name.upper() for name in names]
print(cap_names)
```

```python
# storing the even numbers in the new list from the existing list using the list comprehensions method

numbers = [2,3,45,6,7,5,32]

even_numbers = [i for i in numbers if i%2 == 0]
print(even_numbers)
```
