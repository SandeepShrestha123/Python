```python
'''Walrus operator example of highest marks'''

marks = [10,20,30,40,50,60,70,80]

if (n := max(marks)) >= 50:
    print(f"You have scored the highest marks {n}")

else:
    print(f"You have not scored the highest marks as your marks is {n}")
    
```
