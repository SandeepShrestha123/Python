```python
'''
Write a program to open three files 1.txt, 2.txt and 3.txt. If any of these files are not
present, a message without exiting the program must be printed prompting the same.
'''

files = ["1.txt", "2.txt", "3.txt"]
for file in files:
    try:
        with open(file, "r") as f:
            print(f"\nContents of {file}:")
            print(f.read())
    except FileNotFoundError:
        print(f"{file} was not found.")
```

```python
"""
Write a program to print third, fifth and seventh element from a list using enumerate
function.
"""

numbers = [90, 30, 20, 10 ,5, 1, 2, 7]

print("Printing the numbers only from the index 3, 5 and 7")
for index, items in enumerate(numbers):
    if (index == 3 or index == 5 or index == 7):
        print(f"The number at the index {index} is {items}")
```

```python
"""
Write a list comprehension to print a list which contains the multiplication table of a user
entered number.
"""

num = int(input("Enter a number of which you want to find the multiplication table of: "))

table = [f"{num} x {i} = {num*i}" for i in range(1, 11)]

for line in table:
    print(line)
```

```python
'''
Write a program to display a/b where a and b are integers. If b=0, display infinite by
handling the ‘ZeroDivisionError’ .
'''

def division(a, b):
    try:
        res = a/b
        print(f"The number {a} and {b} after division is {res}.")
    except ZeroDivisionError:
        print("Division cannot be performed as the denominator is 0!")
    else:
        print("division successfull")
        
    
a = int(input("Enter numerator: "))
b = int(input("Enter denominator: "))
division(a,b)
```
