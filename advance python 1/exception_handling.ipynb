```python
'''
Write a program that takes two numbers as input and divides them. Handle the case where the denominator is 0.
'''


def division(a: int, b: int):
    try:
        res = a/b
        return res
    except ZeroDivisionError:
        print("Division cannot be performed when the denominator is zero!")
        
a = int(input("Enter the value of a: "))
b = int(input("Enter the value of b: "))
res = division(a,b)
print(f"The result after the division of two number {a} and {b} is {res}")
        
    
```

```python
'''
Write a program that opens a file named student.txt. If the file does not exist, display an error message instead of crashing.
'''

try:
    with open("example.txt", "r", encoding="UTF-8") as f:
        content = f.read()
        print(content)
except Exception as e:
    print(e)
```

```python
'''
Create a list with five elements. Ask the user to enter an index and print the corresponding element. Handle the case where the index is out of range.
'''

def print_no(datas):
    try:
        a = int(input("Enter the index number: "))
        print(f"The number at index {a} in the list is {datas[a]}")
    except Exception as e:
        print("Enter valid index as the input index is out of range!!")
    
numbers = [1,2,3,4,5]
print_no(numbers)
```

```python
'''
Create a dictionary containing student names and marks. Ask the user for a student's name and print their marks. Handle the case where the name is not found.
'''


def check(mark):
    try:
        n = input("Enter the name of the studnet of which u want to know the marks:  ").lower()
        print(f"The marks of the student {n} is {mark[n]}")
    except KeyError:
        print("The name is not found in the dictionary")


marks = {
    "sandeep": 99,
    "srestaa": 90,
    "shashwot": 90
}
check(marks)

```
